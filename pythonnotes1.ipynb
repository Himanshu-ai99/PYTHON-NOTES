```python
#Python Data Types 

# Numeric
x = 10                  # int
x = 10.5                # float
x = 3 + 4j              # complex

# Boolean
x = True                # bool

# Text
x = "Hello"             # str

# Sequence
x = [1, 2, 3]           # list
x = (1, 2, 3)           # tuple
x = range(5)            # range

# Mapping
x = {"name": "Krish"}   # dict

# Set
x = {1, 2, 3}           # set
x = frozenset([1, 2])   # frozenset

# Binary
x = b"Hello"            # bytes
x = bytearray(b"Hello") # bytearray
x = memoryview(b"Hello")# memoryview

# None
x = None                # NoneType
```

```python
# PYTHON DATA TYPES

# Python has several built-in data types.
#
# 1. Numeric Types
#       int
#       float
#       complex
#
# 2. Boolean Type
#       bool
#
# 3. Sequence Types
#       str
#       list
#       tuple
#       range
#
# 4. Set Types
#       set
#       frozenset
#
# 5. Mapping Type
#       dict
#
# 6. Binary Types
#       bytes
#       bytearray
#       memoryview
#
# 7. None Type
#       NoneType



# 1. INTEGER (int)
#-------------------

age = 18
marks = 95
temperature = -10

print("INTEGER")
print(age)
print(type(age))

# USES:
# - Counting things
# - Age
# - Marks
# - Number of students
# - IDs
# - Mathematical calculations



# 2. FLOAT (float)
#--------------------

price = 99.99
height = 5.8
percentage = 87.5

print("\nFLOAT")
print(price)
print(type(price))

# USES:
# - Decimal numbers
# - Prices
# - Measurements
# - Scientific calculations
# - Percentages



# 3. COMPLEX (complex)
# -----------------------

number = 3 + 4j

print("\nCOMPLEX")
print(number)
print(type(number))

# Access real and imaginary parts
print("Real:", number.real)
print("Imaginary:", number.imag)

# USES:
# - Mathematics
# - Engineering
# - Signal processing
# - Scientific calculations



# 4. BOOLEAN (bool)
# ----------------------

is_student = True
is_logged_in = False

print("\nBOOLEAN")
print(is_student)
print(type(is_student))

# USES:
# - Conditions
# - Decision making
# - Checking True/False situations
#
# Example:

if is_student:
    print("The person is a student")



# 5. STRING (str)
#--------------------

name = "Anshu"
message = "Hello World"

print("\nSTRING")
print(name)
print(type(name))

# String operations
print(name.upper())
print(name.lower())
print(len(name))

# USES:
# - Names
# - Text
# - Messages
# - User input
# - File contents
# - Website data



# 6. LIST (list)
# ---------------------

students = ["Rahul", "Aman", "Krish", "Rohan"]

print("\nLIST")
print(students)
print(type(students))

# Lists can be modified
students.append("Ankit")
students.remove("Aman")

print(students)

# Access an element
print(students[0])

# USES:
# - Store multiple values
# - Student names
# - Products
# - Marks
# - Data that needs to be changed
#
# Lists are:
# - Ordered
# - Mutable
# - Allow duplicate values



# 7. TUPLE (tuple)
# ---------------------

coordinates = (10, 20)
months = ("January", "February", "March")

print("\nTUPLE")
print(coordinates)
print(type(coordinates))

# Access value
print(coordinates[0])

# USES:
# - Data that should not change
# - Coordinates
# - Fixed configuration
# - Returning multiple values from a function
#
# Tuples are:
# - Ordered
# - Immutable
# - Allow duplicate values



# 8. RANGE (range)
# ----------------------

numbers = range(1, 6)

print("\nRANGE")
print(numbers)
print(type(numbers))

print(list(numbers))

# USES:
# - Loops
# - Generating sequences of numbers
# - Repeating an operation
#
# Example:

for i in range(5):
    print(i)



# 9. SET (set)
# ----------------------

numbers = {1, 2, 3, 4, 4, 5}

print("\nSET")
print(numbers)
print(type(numbers))

# Notice that duplicate 4 is removed.

# Add an element
numbers.add(6)

print(numbers)

# USES:
# - Removing duplicates
# - Finding unique values
# - Mathematical set operations
# - Membership checking
#
# Sets are:
# - Unordered
# - Mutable
# - Do not allow duplicate values



# 10. FROZENSET (frozenset)
# -----------------------------

fixed_numbers = frozenset([1, 2, 3, 4])

print("\nFROZENSET")
print(fixed_numbers)
print(type(fixed_numbers))

# USES:
# - Set that cannot be modified
# - Dictionary keys
# - Situations where set data must remain unchanged



# 11. DICTIONARY (dict)
# -----------------------

student = {
    "name": "Himanshu"
    "passed": False
}

print("\nDICTIONARY")
print(student)
print(type(student))

# Access values using keys
print(student["name"])
print(student["marks"])

# Add a new key
student["course"] = "Python"

print(student)

# USES:
# - Store data in key-value pairs
# - Student information
# - User profiles
# - API/JSON data
# - Database-like records
#
# Example:
#
# "name"  -> key
# "Krish" -> value



# 12. BYTES (bytes)
# --------------------

data = b"Hello"

print("\nBYTES")
print(data)
print(type(data))

# USES:
# - Binary data
# - Files
# - Networking
# - Images
# - Audio/video processing
# - Data transmission



# 13. BYTEARRAY (bytearray)
# --------------------------

data = bytearray(b"Hello")

print("\nBYTEARRAY")
print(data)
print(type(data))

# bytearray can be modified
data[0] = 74

print(data)

# USES:
# - Mutable binary data
# - File processing
# - Networking
# - Low-level data manipulation



# 14. MEMORYVIEW (memoryview)
# -------------------------------

data = bytearray("Hello", "utf-8")

view = memoryview(data)

print("\nMEMORYVIEW")
print(view)
print(type(view))

# Convert back to bytes
print(view.tobytes())

# USES:
# - Accessing binary data without copying it
# - Efficient memory usage
# - Large files
# - Image/video processing
# - Performance-sensitive applications



# 15. NONE (NoneType)
# ---------------------

result = None

print("\nNONE")
print(result)
print(type(result))

# USES:
# - Represents absence of a value
# - Functions that don't return anything
# - Default values
# - Checking whether a value exists
#
# Example:

username = None

if username is None:
    print("Username has not been provided")




# SUMMARY
# ============================================================

print("\n================ DATA TYPES SUMMARY ================")

data_types = {
    "int": "Whole numbers",
    "float": "Decimal numbers",
    "complex": "Complex mathematical numbers",
    "bool": "True / False",
    "str": "Text",
    "list": "Ordered, changeable collection",
    "tuple": "Ordered, unchangeable collection",
    "range": "Sequence of numbers",
    "set": "Unique unordered values",
    "frozenset": "Immutable set",
    "dict": "Key-value pairs",
    "bytes": "Immutable binary data",
    "bytearray": "Mutable binary data",
    "memoryview": "View of binary data",
    "NoneType": "Represents no value"
}

for data_type, use in data_types.items():
    print(f"{data_type:12} -> {use}")
```
